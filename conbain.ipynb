```python
def filterCsv(csv, num):
    csv.sort_values(by='f', ascending=False, inplace=True)
    # print(csv[['index', 'con']])
    print(len(csv))
    return csv['index'].head(num)
```

```python
def show(path_list, index = -1, colum = -1, num = -1):
    record = pd.DataFrame()
    csv_list = []
    for p in path_list:
        csv = pd.read_csv(os.path.join(p, 'record.csv'))
        csv['con'] = csv['insert_mean'] - csv['delete_mean']
        csv['index'] = csv['image_id'] + csv['label'].apply(lambda x : str(x))
        csv_list.append(csv)
    if index != -1:
        csv = csv_list[index[0]]
        csv['f'] = csv_list[index[0]][colum] - csv_list[index[1]][colum]
        fil = filterCsv(csv, num)
        for index, csv in enumerate(csv_list):
            p = path_list[index]
            record[p.split('/')[-3] + "_" + p.split('/')[-2]] = \
                csv.loc[csv['index'].isin(fil), ['delete_mean', 'insert_mean', 'con', 'drop', 'increase', 'hits', 'loc']].describe().loc['mean', :]
    else:
        for index, csv in enumerate(csv_list):
            p = path_list[index]
            record[p.split('/')[-3] + "_" + p.split('/')[-2]] = \
                csv.loc[:, ['delete_mean', 'insert_mean', 'con', 'drop', 'increase', 'hits', 'loc']].describe().loc['mean', :]
    return record.T
```

```python
def show_one(path):
    csv = pd.read_csv(os.path.join(path, 'record.csv'))
    return csv
```

```python
def concat(save_path):
    path_list = []
    for p in os.listdir(save_path):
        if p[-3:] == 'csv':
            path_list.append(os.path.join(save_path, p))
    if len(path_list) != 1:
        csv = pd.DataFrame()
        for p in path_list:
            try:
                csv_ = pd.read_csv(p)
                csv_.index = list(range(len(csv_)))
                csv = pd.concat([csv, csv_], ignore_index=True)
                csv.index = list(range(len(csv)))
            except Exception:
                pass
        csv.to_csv(os.path.join(save_path, 'record.csv'), index=False)
```
